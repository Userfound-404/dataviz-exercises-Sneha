```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task 1 — Multi-Series Line Chart with Highlight

**What to build:** A line chart showing CO2 emissions over time for **all Asian countries** in the dataset, with one country highlighted.

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#DDDDDD), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

> 💡 `df[df['Region'] == 'Asia']` to filter; use `go.Figure()` with a loop for per-country control


```python
# Task 1 — Multi-series line with highlight
# YOUR CODE HERE
import pandas as pd
import plotly.graph_objects as go

# Filter to Asia
asia_df = df[df['Region'] == 'Asia']

# Choose highlight country
highlight_country = 'China'

fig = go.Figure()

# Add all countries
for country in asia_df['Country'].unique():
    country_df = asia_df[asia_df['Country'] == country]
    
    if country == highlight_country:
        fig.add_trace(go.Scatter(
            x=country_df['Year'],
            y=country_df['CO2_Mt'],
            mode='lines',
            line=dict(width=3),  # thicker line
            name=country,
        ))
        
        # Direct label at end
        fig.add_trace(go.Scatter(
            x=[country_df['Year'].iloc[-1]],
            y=[country_df['CO2_Mt'].iloc[-1]],
            text=[country],
            mode='text',
            textposition='middle right',
            showlegend=False
        ))
        
    else:
        fig.add_trace(go.Scatter(
            x=country_df['Year'],
            y=country_df['CO2_Mt'],
            mode='lines',
            line=dict(color='#DDDDDD', width=1),
            hoverinfo='skip',
            showlegend=False
        ))

# Layout styling (Lecture 2 rules)
fig.update_layout(
    title="China dominates Asia’s CO₂ growth, pulling far ahead since 2005",
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial'),
    xaxis=dict(
        showgrid=False,
        showline=False,
        title=""
    ),
    yaxis=dict(
        showgrid=False,
        showline=False,
        title="CO₂ emissions (Mt)"
    ),
    showlegend=False
)

fig.show()
```

## Task 2 — Slopegraph: Regional Change 2000 vs 2022

**What to build:** A slopegraph comparing **average regional CO2 emissions** between 2000 and 2022.

**Requirements:**
- One line per region (not per country — aggregate first)
- Colour: regions that increased = one colour; decreased = another
- Values labelled at both ends of each line
- No y-axis tick labels (the endpoint labels make them redundant)
- Insight title stating which regions moved most

> 💡 `df.groupby(['Region','Year'])['CO2_Mt'].mean().reset_index()` then filter to 2000 and 2022


```python
# Task 2 — Slopegraph: regional averages
# YOUR CODE HERE
import pandas as pd
import plotly.graph_objects as go

# Aggregate to regional averages
regional = df.groupby(['Region', 'Year'])['CO2_Mt'].mean().reset_index()

# Keep only 2000 and 2022
slope_df = regional[regional['Year'].isin([2000, 2022])]

# Pivot for easy comparison
pivot = slope_df.pivot(index='Region', columns='Year', values='CO2_Mt').reset_index()
pivot.columns = ['Region', 'y2000', 'y2022']

fig = go.Figure()

for _, row in pivot.iterrows():
    color = '#2ca02c' if row['y2022'] > row['y2000'] else '#d62728'
    
    # Line
    fig.add_trace(go.Scatter(
        x=[0, 1],
        y=[row['y2000'], row['y2022']],
        mode='lines',
        line=dict(color=color, width=2),
        showlegend=False
    ))
    
    # Left label (2000)
    fig.add_trace(go.Scatter(
        x=[0],
        y=[row['y2000']],
        text=[f"{row['Region']} {row['y2000']:.0f}"],
        mode='text',
        textposition='middle left',
        showlegend=False
    ))
    
    # Right label (2022)
    fig.add_trace(go.Scatter(
        x=[1],
        y=[row['y2022']],
        text=[f"{row['y2022']:.0f}"],
        mode='text',
        textposition='middle right',
        showlegend=False
    ))

# Layout styling
fig.update_layout(
    title="Asia shows the largest rise in average CO₂ emissions since 2000, while Europe declines",
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial'),
    xaxis=dict(
        tickvals=[0, 1],
        ticktext=['2000', '2022'],
        showgrid=False,
        showline=False,
        zeroline=False
    ),
    yaxis=dict(
        showgrid=False,
        showline=False,
        showticklabels=False,  # key requirement
        title=""
    )
)

fig.show()
```
