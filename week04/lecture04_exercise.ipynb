## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE
import pandas as pd
import plotly.express as px

df = px.data.gapminder()

# Choose one continent (not Africa)
continent = "Europe"

df_cont = df.loc[df["continent"] == continent]

# Keep only the two years
df_2002_2007 = df_cont[df_cont["year"].isin([2002, 2007])]

fig = px.scatter(
    df_2002_2007,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    size="pop",
    hover_name="country",
    log_x=True,
    title=f"GDP vs Life Expectancy ({continent}: 2002 vs 2007)"
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE
import pandas as pd
import plotly.express as px

df = px.data.gapminder()

df_2007 = df[df["year"] == 2007].copy()

highlight_continent = "Asia"

df_2007["highlight"] = df_2007["continent"].apply(
    lambda x: highlight_continent if x == highlight_continent else "Other"
)
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="highlight",
    hover_name="country",
    log_x=True,
    size_max=60,
    color_discrete_map={
        highlight_continent: "red",
        "Other": "lightgrey"
    },
    title="Global Development in 2007: GDP vs Life Expectancy"
)
fig.add_annotation(
    x=20000,
    y=70,
    text="Asia shows wide variation:<br>high-growth economies vs low-income countries",
    showarrow=True,
    arrowhead=2,
    ax=40,
    ay=-40
)
fig.update_traces(marker=dict(opacity=0.7, line=dict(width=0.5, color="black")))

fig.update_layout(
    xaxis_title="GDP per capita (log scale)",
    yaxis_title="Life expectancy",
    legend_title=""
)

fig.show()
```
